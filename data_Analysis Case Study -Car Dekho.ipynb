```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('car_Dekho_DA.csv')

# Feature Engineering: Calculate Vehicle Age
benchmark_year = 2020  # Max year in dataset is 2018
df['Vehicle_Age'] = benchmark_year - df['Year']

# Set global visual style
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 11, 'figure.autolayout': True})

# ---------------------------------------------------------
# 1. Age vs. Value Retention Plot
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.regplot(
    data=df, x='Vehicle_Age', y='Selling_Price', 
    color='teal', scatter_kws={'alpha': 0.6, 's': 50}, 
    line_kws={'color': 'darkred', 'linewidth': 2}
)
plt.title('1. Vehicle Age vs. Value Retention', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Vehicle Age (Years)', fontweight='bold')
plt.ylabel('Selling Price (Lakhs INR)', fontweight='bold')
plt.show()

# ---------------------------------------------------------
# 2. Fuel Type Dynamics Plot
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=df, x='Fuel_Type', y='Selling_Price', palette='Set2', errorbar=None)
for p in ax.patches:
    ax.annotate(f'₹{p.get_height():.2f}L', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 8), textcoords='offset points', fontweight='bold')

plt.title('2. Fuel Type Dynamics (Average Selling Price)', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Fuel Type', fontweight='bold')
plt.ylabel('Average Selling Price (Lakhs INR)', fontweight='bold')
plt.ylim(0, df.groupby('Fuel_Type')['Selling_Price'].mean().max() * 1.15)
plt.show()

# ---------------------------------------------------------
# 3. Transmission Premium Plot
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df, x='Transmission', y='Selling_Price', palette='Blues', 
    showmeans=True, meanprops={"marker":"o", "markerfacecolor":"red", "markeredgecolor":"red", "markersize":"8"}
)
plt.title('3. Transmission Premium (Price Distribution)', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Transmission Type', fontweight='bold')
plt.ylabel('Selling Price (Lakhs INR)', fontweight='bold')
plt.show()

# ---------------------------------------------------------
# 4. Seller Channel Variance Plot
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=df, x='Seller_Type', y='Selling_Price', palette='magma', errorbar=None)
for p in ax.patches:
    ax.annotate(f'₹{p.get_height():.2f}L', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 8), textcoords='offset points', fontweight='bold')

plt.title('4. Seller Channel Variance (Dealer vs. Individual)', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Seller Channel', fontweight='bold')
plt.ylabel('Average Selling Price (Lakhs INR)', fontweight='bold')
plt.ylim(0, df.groupby('Seller_Type')['Selling_Price'].mean().max() * 1.15)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 1. Load Dataset
df = pd.read_csv('car_Dekho_DA.csv')

# 2. Data Overview & Inspection
print("--- DATASET OVERVIEW ---")
print(f"Dataset Shape: {df.shape}")
print("\nFirst 5 Rows:")
print(df.head())
print("\nDataset Info:")
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())
print("\nSummary Statistics:")
print(df.describe())

# 3. Feature Engineering
# Calculate Age of the vehicle (Assuming current benchmark year as max year + 1 or 2020)
current_year = 2020
df['Vehicle_Age'] = current_year - df['Year']
df['Price_Depreciation'] = df['Present_Price'] - df['Selling_Price']
df['Depreciation_Percentage'] = (df['Price_Depreciation'] / df['Present_Price']) * 100

print("\n--- FEATURE ENGINEERED DATA SAMPLE ---")
print(df[['Car_Name', 'Year', 'Vehicle_Age', 'Selling_Price', 'Present_Price', 'Depreciation_Percentage']].head())

# 4. Data Visualizations

# A. Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Numerical Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# B. Selling Price vs Vehicle Age
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Vehicle_Age', y='Selling_Price', hue='Fuel_Type', s=70)
plt.title('Selling Price vs Vehicle Age by Fuel Type', fontsize=14, fontweight='bold')
plt.xlabel('Vehicle Age (Years)')
plt.ylabel('Selling Price (Lakhs)')
plt.tight_layout()
plt.show()

# C. Average Selling Price by Fuel Type & Transmission
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x='Fuel_Type', y='Selling_Price', hue='Transmission', palette='Set2', ci=None)
plt.title('Average Selling Price by Fuel Type & Transmission', fontsize=14, fontweight='bold')
plt.xlabel('Fuel Type')
plt.ylabel('Average Selling Price (Lakhs)')
plt.tight_layout()
plt.show()

# D. Seller Type vs Selling Price Distribution
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Seller_Type', y='Selling_Price', palette='Pastel1')
plt.title('Selling Price Distribution by Seller Type', fontsize=14, fontweight='bold')
plt.xlabel('Seller Type')
plt.ylabel('Selling Price (Lakhs)')
plt.tight_layout()
plt.show()

# 5. Summary Insights Printing
print("\n--- KEY ANALYSIS SUMMARY ---")
print(f"Average Selling Price: {df['Selling_Price'].mean():.2f} Lakhs")
print(f"Average Present Price: {df['Present_Price'].mean():.2f} Lakhs")
print(f"Average Vehicle Age: {df['Vehicle_Age'].mean():.1f} Years")
print("\nAverage Price by Fuel Type:")
print(df.groupby('Fuel_Type')['Selling_Price'].mean())
print("\nAverage Price by Transmission:")
print(df.groupby('Transmission')['Selling_Price'].mean())
```
